**Adım 1: Yükleme ve ilk bakış**

**Adım 2: Sütun yapısını anlama**

```python
# Sütunları ön eklerine göre gruplandırıyoruz
# filter(regex=...) ile "C1, C2, ..., C14" gibi isimleri tek seferde yakalıyoruz
# ^C\d+$ : C ile başlayıp sadece rakamlarla biten isimler (ProductCD gibi sütunları yanlışlıkla yakalamasın diye)
groups = {
    "C (sayaç)":      df_trans.filter(regex=r"^C\d+$").columns,
    "D (zaman farkı)": df_trans.filter(regex=r"^D\d+$").columns,
    "M (eşleşme T/F)": df_trans.filter(regex=r"^M\d+$").columns,
    "V (Vesta özellikleri)": df_trans.filter(regex=r"^V\d+$").columns,
    "card":           df_trans.filter(regex=r"^card\d+$").columns,
}

for name, cols in groups.items():
    print(f"{name:25s}: {len(cols):3d} sütun")

# Gruplara girmeyen, "tek başına" sütunlar
grouped = set().union(*[set(c) for c in groups.values()])
others = [c for c in df_trans.columns if c not in grouped]
print("\nDiğer sütunlar:", others)
```

```python
# Veri tiplerinin dağılımı: kaç sütun float, int, object (metin)?
print(df_trans.dtypes.value_counts())

# Metin (object) sütunlar kategorik değişkenlerdir ve bellekte en pahalı tip budur
# Hem transaction hem identity için bunları listeliyoruz
obj_trans = df_trans.select_dtypes(include=["object","str"]).columns.tolist()
obj_id = df_id.select_dtypes(include=["object","str"]).columns.tolist()
print("\nTransaction object sütunlar:", obj_trans)
print("Identity object sütunlar:   ", obj_id)
```

```python
# Kategorik sütunların kaç farklı değeri var? (cardinality)
# Az değerli sütunlar "category" tipine çevrilince çok yer kazandırır (Adım 5'te kullanacağız)
# Çok değerli olanlar (ör. DeviceInfo) ise ayrı bir strateji gerektirir
card_trans = df_trans[obj_trans].nunique().sort_values(ascending=False)
card_id = df_id[obj_id].nunique().sort_values(ascending=False)
print(card_trans, "\n")
print(card_id)
```

```python
# Birleştirmeye hazırlık: anahtar sütun (TransactionID) kontrolü
# 1) Transaction'da her ID bir kez mi geçiyor? (unique olmalı)
print("Transaction ID unique mi?:", df_trans["TransactionID"].is_unique)
print("Identity ID unique mi?:   ", df_id["TransactionID"].is_unique)

# 2) Identity'deki ID'lerin hepsi transaction'da var mı?
# isin() ile kontrol ediyoruz; hepsi True ise identity tamamen transaction'ın alt kümesidir
in_trans = df_id["TransactionID"].isin(df_trans["TransactionID"])
print("Identity ID'lerinin transaction'da bulunma oranı:", in_trans.mean())

# 3) Transaction'ların yüzde kaçının kimlik kaydı var?
has_id = df_trans["TransactionID"].isin(df_id["TransactionID"])
print(f"Kimlik kaydı olan işlem oranı: {has_id.mean():.1%}")
```

**Adım 3: Tekrarlı kayıtlar**

```python
# 1) ID hariç tüm sütunlar birebir aynı olan satırlar var mı?
# TransactionID zaten benzersiz olduğu için onu dahil edersek sonuç hep 0 çıkar, o yüzden çıkarıyoruz
# duplicated() varsayılan olarak ilk görüleni "tekrar değil", sonrakileri "tekrar" sayar
cols_wo_id = [c for c in df_trans.columns if c != "TransactionID"]
dup_full = df_trans.duplicated(subset=cols_wo_id)
print("Transaction - ID hariç birebir aynı satır:", dup_full.sum())

# Identity için aynı kontrol
id_cols_wo_id = [c for c in df_id.columns if c != "TransactionID"]
print("Identity - ID hariç birebir aynı satır:   ", df_id.duplicated(subset=id_cols_wo_id).sum())
```

```python
# 3 satırın hangileri olduğunu görelim
# keep=False: tekrar eden grubun TÜM üyelerini işaretler (ilk görüleni de), böylece çiftleri yan yana görürüz
mask = df_trans.duplicated(subset=cols_wo_id, keep=False)

# Aynı gruptaki satırlar alt alta gelsin diye sıralıyoruz
# 3 tekrar var ama grup sayısı 3'ten az olabilir (bir grupta 3 kopya da olabilir)
show_cols = ["TransactionID", "TransactionDT", "TransactionAmt", "card1", "ProductCD", "isFraud"]
df_trans.loc[mask, show_cols].sort_values(["TransactionDT", "card1", "TransactionAmt"])
```

```python
# Her çiftte ilk kaydı tutup ikincisini atacağız
# keep="first": ilk görüleni "tekrar değil", sonrakini "tekrar" sayar
dup_mask = df_trans.duplicated(subset=cols_wo_id, keep="first")
ids_to_drop = df_trans.loc[dup_mask, "TransactionID"]
print("Silinecek ID'ler:", ids_to_drop.tolist())

# Bu ID'lerin identity tablosunda kaydı var mı? (varsa kimlik bilgisi kaybolur)
print("Identity'de kaydı olan:", ids_to_drop.isin(df_id["TransactionID"]).sum())
```

```python
before = len(df_trans)
df_trans = df_trans[~dup_mask].reset_index(drop=True)
print(f"{before:,} -> {len(df_trans):,} satır")   
```

```python
# Silme sonrası tabloda tekrar hesaplıyoruz (satır sayısı değişti)
# keep=False: tekrar eden grubun TÜM üyelerini işaretler
key_cols = ["TransactionDT", "TransactionAmt", "card1", "card2", "card3",
            "card4", "card5", "card6", "ProductCD"]
dup_key = df_trans.duplicated(subset=key_cols, keep=False)
print("Anahtar sütunlarda tekrar eden satır sayısı:", dup_key.sum())
print(f"Oran: {dup_key.mean():.3%}")
```

```python
# Tekrar eden satırları gruplayıp her grubun büyüklüğüne bakıyoruz
# Çoğu grup 2'li olmalı; 5-10'luk büyük gruplar varsa bunlar ayrıca dikkat çekici
dups = df_trans[dup_key]
group_sizes = dups.groupby(key_cols, dropna=False).size()
print("Grup sayısı:", len(group_sizes))
print(group_sizes.value_counts().sort_index())   # kaç tane 2'li, 3'lü... grup var

# Bu satırlarda isFraud dağılımı: bilgi amaçlı, analiz değil
# Genel fraud oranıyla (~%3,5) kıyaslayıp EDA için not alacağız
print("\nTekrar eden satırlarda isFraud:")
print(dups["isFraud"].value_counts())
```

```python
dup_id = df_id.duplicated(subset=id_cols_wo_id, keep=False)
filled = df_id[id_cols_wo_id].notna().sum(axis=1)

print("Tekrar eden satırlarda ortalama dolu sütun:  ", round(filled[dup_id].mean(), 1))
print("Tekrar etmeyen satırlarda ortalama dolu sütun:", round(filled[~dup_id].mean(), 1))

top_dups = (df_id[dup_id]
            .groupby(["DeviceType", "DeviceInfo", "id_31"], dropna=False)
            .size()
            .sort_values(ascending=False)
            .head(10))
print(top_dups)
```

**Adım 4: Boşluk analizi**

```python
# Her sütunun boşluk oranını (%) hesaplayan yardımcı fonksiyon
# isna().mean(): her sütunda NaN olan satırların oranı (True=1 sayıldığı için ortalaması oranı verir)
def missing_table(df):
    miss = df.isna().mean() * 100
    # Sadece boşluğu olanları, büyükten küçüğe sıralı göster
    return miss[miss > 0].sort_values(ascending=False).round(2)

miss_trans = missing_table(df_trans)
miss_id = missing_table(df_id)

print(f"Transaction: {len(miss_trans)} / {df_trans.shape[1]} sütunda boşluk var")
print(f"Identity:    {len(miss_id)} / {df_id.shape[1]} sütunda boşluk var")
```

```python
# Boşluk oranlarını aralıklara ayırıp kaç sütunun hangi aralıkta olduğuna bakalım
# pd.cut: sürekli değerleri (oranları) önceden belirlediğimiz kovalara dağıtır
bins = [0, 5, 25, 50, 75, 100]
labels = ["0-5%", "5-25%", "25-50%", "50-75%", "75-100%"]

# Boşluğu hiç olmayan sütunlar da dahil olsun diye tüm sütunların oranını alıyoruz
all_trans = df_trans.isna().mean() * 100
all_id = df_id.isna().mean() * 100

# include_lowest=True: 0% olan sütunlar da ilk kovaya girsin
dist = pd.DataFrame({
    "Transaction": pd.cut(all_trans, bins=bins, labels=labels, include_lowest=True).value_counts(),
    "Identity":    pd.cut(all_id, bins=bins, labels=labels, include_lowest=True).value_counts(),
}).reindex(labels)
dist
```

```python
# Önceki adımda oluşturduğumuz gruplara göre ortalama boşluk oranı
# Böylece "V sütunları ne kadar boş?" sorusuna tek satırda cevap veriyoruz
# groups sözlüğü Adım 2'den geliyor; bellekte yoksa o hücreyi tekrar çalıştır
for name, cols in groups.items():
    print(f"{name:25s}: ortalama boşluk %{all_trans[cols].mean():.1f}")

# "Diğer" sütunlar için tek tek bakalım, çünkü bunlar analizin ana malzemesi
print("\nDiğer sütunlar:")
print(all_trans[others].round(2).sort_values(ascending=False))
```

```python
# %75'ten fazla boş olan sütunların hangi gruplara dağıldığına bakalım
# all_trans: önceki bloktaki her sütunun boşluk oranı (%)
very_empty = all_trans[all_trans > 75].index

# Her grup için: kaç sütunun %75'ten fazla boş olduğu ve grubun toplam sütun sayısı
# groups sözlüğü Adım 2'den geliyor
for name, cols in groups.items():
    n_empty = len(set(cols) & set(very_empty))
    print(f"{name:25s}: {n_empty:3d} / {len(cols):3d} sütun >%75 boş")

# Gruplara girmeyen sütunlardan (others) kaçı çok boş?
print("\nDiğer sütunlardan çok boş olanlar:",
      [c for c in others if c in very_empty])
```

```python
# addr1 ve addr2 aynı satırlarda mı boş?
# (a & b).sum(): ikisi birden boş olan satır sayısı; (a | b).sum(): en az biri boş olan satır sayısı
# Eşitse, ikisi tam olarak aynı satırlarda boş demektir
a1, a2 = df_trans["addr1"].isna(), df_trans["addr2"].isna()
print("İkisi birden boş:", (a1 & a2).sum(), "| en az biri boş:", (a1 | a2).sum())

# Boşluksuz sütunlar: tahminimizi doğrulayalım
complete_trans = df_trans.columns[df_trans.isna().sum() == 0].tolist()
complete_id = df_id.columns[df_id.isna().sum() == 0].tolist()
print("\nTransaction'da boşluksuz:", len(complete_trans), complete_trans)
print("Identity'de boşluksuz:   ", complete_id)
```

**Adım 5: Bellek optimizasyonu**

```python
# DataFrame'in bellek kullanımını MB olarak döndüren yardımcı fonksiyon
# deep=True: string sütunlarının gerçek boyutunu da hesaplar
def mem_mb(df):
    return df.memory_usage(deep=True).sum() / 1024**2
```

```python
# Dönüşümden önce bir karşılaştırma örneği saklıyoruz
# TransactionAmt'ı float32'ye çevirince değer bozuluyor mu, bunu sonra kontrol edeceğiz
orig_amt = df_trans["TransactionAmt"].copy()

def reduce_mem(df):
    """Her sütunu tipine göre en küçük güvenli tipe çevirir."""
    for col in df.columns:
        dtype = df[col].dtype

        if pd.api.types.is_float_dtype(dtype):
            # float64 -> float32: hücre başına 8 byte yerine 4 byte
            # float16'ya inmiyoruz: V sütunlarında büyük değerler var ve hassasiyet kaybı riskli
            df[col] = df[col].astype("float32")

        elif pd.api.types.is_integer_dtype(dtype):
            # downcast="integer": değer aralığına sığan en küçük int tipini seçer
            # (isFraud -> int8, card1 -> int16, TransactionID -> int32 gibi)
            # Bu dönüşüm değer kaybettirmez, çünkü pandas aralığı kontrol eder
            df[col] = pd.to_numeric(df[col], downcast="integer")

        else:
            # Kalan her şey metin sütunu (14 + 17 tane)
            # category: her benzersiz değer bir kez saklanır, satırlar küçük kod tutar
            df[col] = df[col].astype("category")
    return df

before_t, before_i = mem_mb(df_trans), mem_mb(df_id)
df_trans = reduce_mem(df_trans)
df_id = reduce_mem(df_id)
after_t, after_i = mem_mb(df_trans), mem_mb(df_id)

print(f"Transaction: {before_t:,.1f} MB -> {after_t:,.1f} MB ({1 - after_t/before_t:.1%} azalma)")
print(f"Identity:    {before_i:,.1f} MB -> {after_i:,.1f} MB ({1 - after_i/before_i:.1%} azalma)")
```

```python
# Doğrulama 1: TransactionAmt'ta float32 dönüşümü ne kadar sapma yarattı?
# Sapma kuruşun çok altındaysa sorun yok
diff = (df_trans["TransactionAmt"].astype("float64") - orig_amt).abs().max()
print("TransactionAmt en büyük sapma:", diff)

# Doğrulama 2: yeni tipler
print(df_trans.dtypes.value_counts())
print(df_trans[["TransactionID", "isFraud", "TransactionDT", "card1"]].dtypes)
```

**Adım 6: Birleştirme ve kaydetme**

```python
# Anahtar sütunun tipi iki tabloda aynı mı? (ikisi de int32 olmalı)
print(df_trans["TransactionID"].dtype, df_id["TransactionID"].dtype)

# left join: transaction tablosunu ana tablo alıp identity'yi üzerine ekliyoruz
# validate="one_to_one": ID'ler her iki tabloda benzersiz olmalı, değilse pandas hata verir
# (Adım 2'de unique olduklarını doğrulamıştık, bu ikinci bir güvence)
df = df_trans.merge(df_id, on="TransactionID", how="left", validate="one_to_one")

# Doğrulamalar: assert koşul False ise hata fırlatır, True ise sessiz geçer
assert len(df) == 590_537, "Satır sayısı değişti!"        # left join satır eklememeli/silmemeli
assert df.shape[1] == 394 + 41 - 1, "Sütun sayısı beklenenden farklı!"   # TransactionID ortak, bir kez sayılır
print("Birleştirme sonrası shape:", df.shape)
```

```python
# Kimlik kaydı olan işlem oranı hâlâ %24,4 mü? (id_01 sadece identity'den geliyor)
# id_01 dolu olan satırlar = identity ile eşleşen işlemler
print(f"Identity ile eşleşen işlem oranı: {df['id_01'].notna().mean():.1%}")

# Birleştirme sonrası bellek ve tip kontrolü
print(f"Birleşik tablo: {mem_mb(df):,.1f} MB")
print(df.dtypes.astype(str).value_counts())
```

```python
# data/ klasörüne kaydediyoruz (.gitignore'da olduğundan emin ol, dosya yüzlerce MB olabilir)
OUT_PATH = "../data/train_merged.parquet"
df.to_parquet(OUT_PATH, index=False)

# Doğrulama: dosyayı geri okuyup tiplerin ve boyutun korunduğunu kontrol et
df_check = pd.read_parquet(OUT_PATH)
print("Okunan shape:", df_check.shape)
print("Tipler aynı mı?:", (df.dtypes == df_check.dtypes).all())
print(f"Bellekte: {mem_mb(df_check):,.1f} MB")

import os
print(f"Dosya boyutu: {os.path.getsize(OUT_PATH) / 1024**2:,.1f} MB")
```

# 01 – Data Overview: Özet ve Bulgular

Bu notebook'un amacı, analize başlamadan önce veriyi **tanımak, temizlik açısından kontrol etmek, küçültmek ve tek bir dosyada toplamak**.

---

## 1. Veri seti nedir?

**IEEE-CIS Fraud Detection** (Kaggle yarışması, veriyi sağlayan şirket: Vesta). Online kart işlemlerinden oluşuyor ve her işlem için "dolandırıcılık mı (`isFraud = 1`), değil mi (`isFraud = 0`)" bilgisi var.

Veri iki ayrı dosyada geliyor:

| Dosya | İçerik | Satır × Sütun | Bellek |
|---|---|---|---|
| `train_transaction.csv` | İşlemin kendisi: tutar, kart, adres, e-posta, ürün tipi, anonim özellikler | 590.540 × 394 | 1.791,7 MB |
| `train_identity.csv` | İşlemi yapan cihaz/tarayıcı bilgisi | 144.233 × 41 | 56,5 MB |

İki dosya `TransactionID` sütunuyla birbirine bağlanıyor.

## 2. Sütunlar neyi anlatıyor?

**Transaction tablosu (394 sütun):**

| Grup | Sütun sayısı | Anlamı |
|---|---|---|
| `TransactionID` | 1 | İşlemin benzersiz numarası |
| `isFraud` | 1 | Hedef değişken (0 = normal, 1 = fraud) |
| `TransactionDT` | 1 | Belirsiz bir başlangıç anından itibaren geçen süre (saniye). **Gerçek bir tarih değil.** |
| `TransactionAmt` | 1 | İşlem tutarı |
| `ProductCD` | 1 | Ürün tipi (5 farklı değer) |
| `card1`–`card6` | 6 | Kart bilgileri (kart tipi, banka vb.) |
| `addr1`, `addr2` | 2 | Adres bilgileri |
| `dist1`, `dist2` | 2 | Mesafe bilgileri |
| `P_emaildomain`, `R_emaildomain` | 2 | Alıcı / gönderici e-posta alan adları |
| `C1`–`C14` | 14 | Sayaç türü özellikler (anlamları gizli) |
| `D1`–`D15` | 15 | Zaman farkı türü özellikler (anlamları gizli) |
| `M1`–`M9` | 9 | Eşleşme (T/F) bayrakları |
| `V1`–`V339` | 339 | Vesta'nın kendi ürettiği özellikler (anlamları gizli) |

**Identity tablosu (41 sütun):** `TransactionID`, `id_01`–`id_38` (anonim cihaz/ağ/tarayıcı özellikleri), `DeviceType` (mobil/masaüstü), `DeviceInfo` (cihaz modeli).

> **Sınırlılık:** Yarışma verisi gizlilik için anonimleştirilmiş. C, D, M, V ve `id_*` sütunlarının gerçek anlamı bilinmiyor. Gerçek bir fintech'te bu sütunların ne olduğunu bilirdik; burada bunlar "anlamı bilinmeyen ham sinyal" olarak ele alınıyor. Verinin %86'sı (339/394) V sütunlarından oluşuyor.

**Veri tipleri (transaction, ham hali):** 376 sütun `float64`, 14 sütun metin, 4 sütun `int64`. Bazı tam sayı olması gereken sütunlar (`card2`, `addr1` gibi) boşluk içerdikleri için ondalıklı okundu.

## 3. İki tablonun ilişkisi

`TransactionID` üzerinden yapılan kontroller:

- Her iki tabloda da `TransactionID` **benzersiz** (bir işlem tek satır).
- Identity'deki ID'lerin **%100'ü** transaction tablosunda var (kimliği olup işlemi olmayan kayıt yok).
- İşlemlerin yalnızca **%24,4'ünün** kimlik (cihaz) kaydı var, yaklaşık dörtte üçünün yok.

**Birleştirme:** Birleştirmede `left join` (transaction ana tablo). `inner join` işlemlerin %75,6'sını silerdi. Ayrıca "kimlik kaydı olmaması" başlı başına bir bilgi olabilir; EDA'da fraud oranıyla ilişkisine bakılacak.

## 4. Tekrarlı kayıt analizi

**a) Transaction'da birebir aynı satırlar**
- `TransactionID` hariç 393 sütunun hepsi aynı olan **3 çift** bulundu.
- Üç çiftte de ID'ler ardışık (örn. 3282456/3282457) ve `TransactionDT` aynı (aynı saniye); üçü de `isFraud = 0`.
- Yorum: Sistemin aynı işlemi iki kez kaydetmesi gibi görünüyor. Bu bir yorum; kesin kanıt değil.
- Her çiftin ikinci kopyası silindi (3 satır). Silinen ID'lerin identity kaydı olmadığı için kimlik bilgisi kaybı yok. Satır sayısı **590.540 → 590.537**.

**b) Identity'de "tekrar eden" satırlar**
- ID hariç diğer sütunları aynı olan **3.767 satır** var.
- Tekrar eden satırlarda ortalama **8,3 sütun** dolu; tekrar etmeyenlerde **25,9**. Yani tekrarlar çoğunlukla bilgisi az (çok boş) kayıtlar. Geri kalanı yaygın cihaz/tarayıcı kombinasyonları (aynı cihazdan birden fazla işlem).
- Her satır farklı bir işleme ait; bunlar gerçek kopya değil. Bu nedenle silinmedi.

**c) "İş mantığı" tekrarı (aynı zaman + tutar + kart + ürün)**
- **102 satır** (%0,017), **41 grup** (31 ikili, 5 üçlü, 3 dörtlü, 1 beşli, 1 sekizli).
- Bu satırlar diğer sütunlarda farklı, yani birebir kopya değil.
- Bu 102 satırın **17'si fraud (%16,7)**.
- EDA'da incelenecek ve feature engineering için aday bir sinyal ("aynı kartın kısa sürede aynı tutarı tekrar denemesi").

## 6. Bellek optimizasyonu

Birleştirmeden önce her tablo ayrı ayrı küçültüldü:

| Dönüşüm | Neden |
|---|---|
| `float64` → `float32` (376 + 23 sütun) | Hücre başına 8 yerine 4 byte. `float16`'ya inilmedi: hassasiyet kaybı riski var. |
| Tam sayılar → en küçük uygun `int` | `isFraud` → `int8`, `card1` → `int16`, `TransactionID`/`TransactionDT` → `int32`. Değer kaybı yok. |
| Metin → `category` (14 + 17 sütun) | Kardinalite düşük (en yüksek: `DeviceInfo` 1.786 farklı değer). Her benzersiz değer bir kez saklanır. |

**Sonuç**

| Tablo | Önce | Sonra | Azalma |
|---|---|---|---|
| Transaction | 1.791,7 MB | 861,1 MB | %51,9 |
| Identity | 56,5 MB | 16,0 MB | %71,7 |

**Doğrulama:** `TransactionAmt` için `float32` dönüşümünün yarattığı en büyük sapma **0,000375**, yani tutar analizi için ihmal edilebilir. Kalan boyutun neredeyse tamamı 376 `float32` sütunundan geliyor.

- `left join`, `TransactionID` üzerinden, `validate="one_to_one"` ile (ID'ler benzersiz olmalı, değilse hata verir).
- Doğrulamalar: satır sayısı **590.537** (değişmedi), sütun sayısı **434** (394 + 41 − 1 ortak ID). Identity ile eşleşen işlem oranı hâlâ **%24,4**.

CSV yerine Parquet seçildi çünkü CSV tip bilgisini saklamaz (`float32` ve `category` tekrar okununca eski haline döner, optimizasyon boşa gider). Parquet tipleri korur, daha küçük ve çok daha hızlı okunur. Dosya geri okunup tiplerin aynı olduğu doğrulandı.
